```python
import numpy as np
import openrouteservice
import folium
import random
import math
from itertools import combinations

API_KEY = '5b3ce3597851110001cf624844e5b52282f74ade88e3c2a308b976e5'  # ← Pega tu API KEY real aquí

MAX_CAPACITY = 500
NUM_ANTS = 30
NUM_ITER = 100
ALPHA = 1  # importancia de feromonas
BETA = 3   # importancia de visibilidad (1/distance)
EVAPORATION = 0.5
Q = 100

locations = [
    (10.983724, -74.789957),  # Origen
    (10.9700532, -74.8019507), (10.9525577, -74.8039492), (10.975318, -74.808005),
    (10.9066255, -74.7856385), (10.9724651, -74.8060456), (11.017676, -74.809716),
    (10.944379, -74.803011), (10.9124556, -74.7826908), (10.9606144, -74.8028657),
    (10.9230761, -74.8008353), (10.9628562, -74.8349773), (10.940902, -74.771705),
    (10.9942585, -74.8122456), (10.963991, -74.817237), (11.027218, -74.869495),
    (10.9615811, -74.8302943), (10.979663, -74.80365), (10.9992743, -74.7924869),
    (10.9625489, -74.830824), (10.985877, -74.83556), (11.024694, -74.869555),
    (10.932394, -74.766357), (11.023623, -74.80664), (10.9030214, -74.7919232),
    (11.02425, -74.86800), (10.9558219, -74.8218484), (10.9599135, -74.8071882),
    (10.894661, -74.885339), (10.9590029, -74.796548), (10.7465427, -74.7565432),
    (11.0218788, -74.8705983),
]
demands = [0, 107, 112, 95, 84, 72, 107, 52, 110, 49, 67, 75, 107, 111, 107, 80, 87, 94, 78, 66, 30,
           70, 69, 94, 70, 114, 64, 46, 76, 33, 60, 69]

# Distancia real con OpenRouteService
client = openrouteservice.Client(key=API_KEY)
coords = [(lon, lat) for lat, lon in locations]
matrix = client.distance_matrix(locations=coords, profile='driving-car',
                                metrics=['distance'], resolve_locations=True, units='m')
dist_matrix = np.array(matrix['distances'])

n = len(locations)
pheromones = np.ones((n, n))
visibility = 1 / (dist_matrix + 1e-10)

def construct_solution():
    unvisited = set(range(1, n))
    routes = []
    while unvisited:
        current_route = [0]
        load = 0
        while True:
            last = current_route[-1]
            candidates = [i for i in unvisited if demands[i] + load <= MAX_CAPACITY]
            if not candidates:
                break
            probs = []
            for j in candidates:
                tau = pheromones[last][j] ** ALPHA
                eta = visibility[last][j] ** BETA
                probs.append(tau * eta)
            probs = np.array(probs)
            probs /= probs.sum()
            next_city = np.random.choice(candidates, p=probs)
            current_route.append(next_city)
            load += demands[next_city]
            unvisited.remove(next_city)
        routes.append(current_route)
    return routes

def total_distance(routes):
    total = 0
    for route in routes:
        for i in range(len(route)-1):
            total += dist_matrix[route[i]][route[i+1]]
    return total

best_routes = None
best_cost = float('inf')

for iteration in range(NUM_ITER):
    all_routes = []
    all_costs = []
    for _ in range(NUM_ANTS):
        solution = construct_solution()
        cost = total_distance(solution)
        all_routes.append(solution)
        all_costs.append(cost)
        if cost < best_cost and all(len(r) > 1 for r in solution):
            best_cost = cost
            best_routes = solution
    pheromones *= (1 - EVAPORATION)
    for routes, cost in zip(all_routes, all_costs):
        for route in routes:
            for i in range(len(route) - 1):
                pheromones[route[i]][route[i+1]] += Q / cost

# Mostrar resultados
print(f"Distancia total: {best_cost / 1000:.2f} km")
print(f"Número de vehículos: {len(best_routes)}")

# === Mapa ===
mapa = folium.Map(location=locations[0], zoom_start=12)
colors = ['red', 'blue', 'green', 'purple', 'orange', 'darkred', 'cadetblue', 'darkgreen']

for i, ruta in enumerate(best_routes):
    coord_ruta = [locations[n] for n in ruta]
    folium.PolyLine(coord_ruta, color=colors[i % len(colors)], weight=4, opacity=0.7).add_to(mapa)
    for n in ruta:
        folium.Marker(locations[n], popup=f"Punto {n} - {demands[n]}kg").add_to(mapa)

mapa.save("rutas_ACO.html")

# === Enlaces Google Maps ===
def generar_links(rutas, locs):
    links = []
    for ruta in rutas:
        pts = [f"{locs[i][0]},{locs[i][1]}" for i in ruta]
        links.append("https://www.google.com/maps/dir/" + "/".join(pts))
    return links

links = generar_links(best_routes, locations)
for i, link in enumerate(links):
    print(f"Ruta {i+1}: {link}")
```
