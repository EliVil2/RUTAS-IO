```python
import numpy as np
import openrouteservice
import random
import urllib.parse

# === Parámetros ===
API_KEY = '5b3ce3597851110001cf624844e5b52282f74ade88e3c2a308b976e5'  # Reemplaza aquí con tu API key válida
MAX_CAPACITY = 500
ALPHA = 1
BETA = 3
EVAPORATION = 0.5
Q = 100
NUM_ANTS = 50
NUM_ITER = 200

# === Datos ===
locations = [
    (10.983724, -74.789957),  # Origen
    (10.9700532, -74.8019507), (10.9525577, -74.8039492), (10.975318, -74.808005),
    (10.9066255, -74.7856385), (10.9724651, -74.8060456), (11.017676, -74.809716),
    (10.944379, -74.803011), (10.9124556, -74.7826908), (10.9606144, -74.8028657),
    (10.9230761, -74.8008353), (10.9628562, -74.8349773), (10.940902, -74.771705),
    (10.9942585, -74.8122456), (10.963991, -74.817237), (11.027218, -74.869495),
    (10.9615811, -74.8302943), (10.979663, -74.80365), (10.9992743, -74.7924869),
    (10.9625489, -74.830824), (10.985877, -74.83556), (11.024694, -74.869555),
    (10.932394, -74.766357), (11.023623, -74.80664), (10.9030214, -74.7919232),
    (11.02425, -74.86800), (10.9558219, -74.8218484), (10.9599135, -74.8071882),
    (10.894661, -74.885339), (10.9590029, -74.796548), (10.7465427, -74.7565432),
    (11.0218788, -74.8705983),
]
demands = [0, 107, 112, 95, 84, 72, 107, 52, 110, 49, 67, 75, 107, 111, 107, 80, 87, 94, 78, 66, 30,
           70, 69, 94, 70, 114, 64, 46, 76, 33, 60, 69]

# === Cliente OpenRouteService ===
client = openrouteservice.Client(key=API_KEY)
coords = [(lon, lat) for lat, lon in locations]

# Obtener matriz de distancias reales (en metros)
matrix = client.distance_matrix(coords, profile='driving-car', metrics=['distance'], units='m')
dist_matrix = np.array(matrix['distances'])
n = len(locations)

# Inicializar feromonas y visibilidad
pheromones = np.ones((n, n))
visibility = 1 / (dist_matrix + 1e-10)

# Funciones auxiliares

def route_distance(route):
    dist = 0
    for i in range(len(route)-1):
        dist += dist_matrix[route[i]][route[i+1]]
    return dist

def total_distance(routes):
    return sum(route_distance(route) for route in routes)

def two_opt(route):
    best = route
    improved = True
    while improved:
        improved = False
        for i in range(1, len(best) - 2):
            for j in range(i+1, len(best)):
                if j - i == 1:
                    continue
                new_route = best[:i] + best[i:j][::-1] + best[j:]
                if route_distance(new_route) < route_distance(best):
                    best = new_route
                    improved = True
        route = best
    return best

def construct_solution():
    unvisited = set(range(1, n))
    routes = []
    while unvisited:
        route = [0]  # Inicio en el origen
        current_load = 0
        current_city = 0
        while True:
            candidates = [j for j in unvisited if demands[j] + current_load <= MAX_CAPACITY]
            if not candidates:
                break
            probs = []
            for j in candidates:
                tau = pheromones[current_city][j] ** ALPHA
                eta = visibility[current_city][j] ** BETA
                probs.append(tau * eta)
            probs = np.array(probs)
            probs /= probs.sum()
            next_city = np.random.choice(candidates, p=probs)
            route.append(next_city)
            current_load += demands[next_city]
            unvisited.remove(next_city)
            current_city = next_city
        # NO se regresa al origen
        routes.append(route)
    # Si hay menos de 5 rutas, agrega rutas vacías para cumplir mínimo 5 vehículos
    while len(routes) < 5:
        routes.append([0])
    return routes

def create_google_maps_link(route):
    points = []
    for idx in route:
        lat, lon = locations[idx]
        points.append(f"{lat},{lon}")
    encoded_points = [urllib.parse.quote(p) for p in points]
    return "https://www.google.com/maps/dir/" + "/".join(encoded_points)

# Algoritmo Colonia de Hormigas
best_routes = None
best_cost = float('inf')

for iteration in range(NUM_ITER):
    all_solutions = []
    all_costs = []
    for _ in range(NUM_ANTS):
        solution = construct_solution()
        # Mejorar cada ruta con 2-opt
        solution = [two_opt(r) for r in solution if len(r) > 1]
        cost = total_distance(solution)
        if cost < best_cost:
            best_cost = cost
            best_routes = solution
        all_solutions.append(solution)
        all_costs.append(cost)
    # Actualizar feromonas
    pheromones *= (1 - EVAPORATION)
    for solution, cost in zip(all_solutions, all_costs):
        for route in solution:
            for i in range(len(route)-1):
                pheromones[route[i]][route[i+1]] += Q / cost
    if iteration % 10 == 0 or iteration == NUM_ITER-1:
        print(f"Iter {iteration+1}/{NUM_ITER} - Mejor distancia: {best_cost/1000:.2f} km")

# Mostrar resultados
print(f"\n🚚 Vehículos usados: {len(best_routes)}")
print(f"📏 Distancia total mínima encontrada: {best_cost/1000:.2f} km")

for i, route in enumerate(best_routes, 1):
    carga = sum(demands[idx] for idx in route)
    print(f"\nRuta {i} → carga: {carga} kg → puntos: {route}")
    print(f"Google Maps: {create_google_maps_link(route)}")
```
