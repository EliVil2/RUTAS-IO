```python
import numpy as np
import openrouteservice
import random
import urllib.parse
import copy

# === Parámetros ===
API_KEY = '5b3ce3597851110001cf624844e5b52282f74ade88e3c2a308b976e5'  # ← ← ← REEMPLAZA ESTO
MAX_CAPACITY = 500
ALPHA = 1
BETA = 3
EVAPORATION = 0.4
Q = 100
NUM_ANTS = 50
NUM_ITER = 200
MIN_VEHICLES = 5

# === Datos ===
locations = [
    (10.983724, -74.789957),  # Origen
    (10.9700532, -74.8019507), (10.9525577, -74.8039492), (10.975318, -74.808005),
    (10.9066255, -74.7856385), (10.9724651, -74.8060456), (11.017676, -74.809716),
    (10.944379, -74.803011), (10.9124556, -74.7826908), (10.9606144, -74.8028657),
    (10.9230761, -74.8008353), (10.9628562, -74.8349773), (10.940902, -74.771705),
    (10.9942585, -74.8122456), (10.963991, -74.817237), (11.027218, -74.869495),
    (10.9615811, -74.8302943), (10.979663, -74.80365), (10.9992743, -74.7924869),
    (10.9625489, -74.830824), (10.985877, -74.83556), (11.024694, -74.869555),
    (10.932394, -74.766357), (11.023623, -74.80664), (10.9030214, -74.7919232),
    (11.02425, -74.86800), (10.9558219, -74.8218484), (10.9599135, -74.8071882),
    (10.894661, -74.885339), (10.9590029, -74.796548), (10.7465427, -74.7565432),
    (11.0218788, -74.8705983),
]
demands = [0, 107, 112, 95, 84, 72, 107, 52, 110, 49, 67, 75, 107, 111, 107, 80, 87, 94, 78, 66, 30,
           70, 69, 94, 70, 114, 64, 46, 76, 33, 60, 69]

# === Cliente y matriz de distancias ===
client = openrouteservice.Client(key=API_KEY)
coords = [(lon, lat) for lat, lon in locations]
matrix = client.distance_matrix(coords, profile='driving-car', metrics=['distance'], units='m')
dist_matrix = np.array(matrix['distances'])
n = len(locations)
pheromones = np.ones((n, n))
visibility = 1 / (dist_matrix + 1e-10)

# === Funciones auxiliares ===
def route_distance(route):
    return sum(dist_matrix[route[i]][route[i+1]] for i in range(len(route)-1))

def total_distance(routes):
    return sum(route_distance(r) for r in routes if len(r) > 1)

def two_opt(route):
    best = route[:]
    improved = True
    while improved:
        improved = False
        for i in range(1, len(best)-2):
            for j in range(i+1, len(best)-1):
                if j - i == 1: continue
                new_route = best[:i] + best[i:j][::-1] + best[j:]
                if route_distance(new_route) < route_distance(best):
                    best = new_route
                    improved = True
    return best

def swap_between_routes(routes):
    best = copy.deepcopy(routes)
    improved = True
    while improved:
        improved = False
        for i in range(len(best)):
            for j in range(i+1, len(best)):
                for idx1 in range(1, len(best[i])):
                    for idx2 in range(1, len(best[j])):
                        a, b = best[i][idx1], best[j][idx2]
                        cap_i = sum(demands[k] for k in best[i]) - demands[a] + demands[b]
                        cap_j = sum(demands[k] for k in best[j]) - demands[b] + demands[a]
                        if cap_i <= MAX_CAPACITY and cap_j <= MAX_CAPACITY:
                            best[i][idx1], best[j][idx2] = b, a
                            if total_distance(best) < total_distance(routes):
                                routes = copy.deepcopy(best)
                                improved = True
                            else:
                                best[i][idx1], best[j][idx2] = a, b
    return routes

def relocate_between_routes(routes):
    best = copy.deepcopy(routes)
    improved = True
    while improved:
        improved = False
        for i in range(len(best)):
            for j in range(len(best)):
                if i == j: continue
                for idx in range(1, len(best[i])):
                    node = best[i][idx]
                    for pos in range(1, len(best[j])):
                        new_i = best[i][:idx] + best[i][idx+1:]
                        new_j = best[j][:pos] + [node] + best[j][pos:]
                        cap_i = sum(demands[k] for k in new_i)
                        cap_j = sum(demands[k] for k in new_j)
                        if cap_i <= MAX_CAPACITY and cap_j <= MAX_CAPACITY:
                            test = copy.deepcopy(best)
                            test[i] = new_i
                            test[j] = new_j
                            if total_distance(test) < total_distance(routes):
                                routes = test
                                improved = True
    return routes

def construct_solution():
    unvisited = set(range(1, n))
    routes = []
    while unvisited:
        route = [0]
        current_load = 0
        current = 0
        while True:
            candidates = [j for j in unvisited if demands[j] + current_load <= MAX_CAPACITY]
            if not candidates: break
            probs = []
            for j in candidates:
                tau = pheromones[current][j] ** ALPHA
                eta = visibility[current][j] ** BETA
                probs.append(tau * eta)
            probs = np.array(probs)
            probs /= probs.sum()
            next_city = np.random.choice(candidates, p=probs)
            route.append(next_city)
            current_load += demands[next_city]
            unvisited.remove(next_city)
            current = next_city
        routes.append(route)
    while len(routes) < MIN_VEHICLES:
        routes.append([0])
    return routes

def create_google_maps_link(route):
    points = []
    for idx in route:
        lat, lon = locations[idx]
        points.append(f"{lat},{lon}")
    return "https://www.google.com/maps/dir/" + "/".join([urllib.parse.quote(p) for p in points])

# === Ejecución del ACO con mejoras ===
best_routes = None
best_cost = float('inf')

for it in range(NUM_ITER):
    for _ in range(NUM_ANTS):
        sol = construct_solution()
        sol = [two_opt(r) for r in sol if len(r) > 1]
        sol = swap_between_routes(sol)
        sol = relocate_between_routes(sol)
        cost = total_distance(sol)
        if cost < best_cost:
            best_cost = cost
            best_routes = copy.deepcopy(sol)
    pheromones *= (1 - EVAPORATION)
    for r in best_routes:
        for i in range(len(r)-1):
            pheromones[r[i]][r[i+1]] += Q / best_cost
    if it % 10 == 0 or it == NUM_ITER - 1:
        print(f"Iteración {it+1}/{NUM_ITER} - Distancia total: {best_cost/1000:.2f} km")

# === Mostrar resultado final ===
print(f"\n🚚 Vehículos usados: {len(best_routes)}")
print(f"📏 Distancia total mínima encontrada: {best_cost/1000:.2f} km")

for i, r in enumerate(best_routes, 1):
    carga = sum(demands[x] for x in r)
    print(f"\nRuta {i} (Carga: {carga} kg): {r}")
    print("Google Maps:", create_google_maps_link(r))
```
